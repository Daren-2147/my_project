Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def is_an_amstrong_number(n):
    total = 0
    hat = len(str(n))
    temp = n
    while n!=0:
        total+=(n%10)**hat
        n//=10
    if total==temp:
        return True
    return False
n = int(input())
print(is_an_amstrong_number(n))

```

VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.


```python
def is_magic_number(n):
    # Xử lý trên dữ liệu kiểu số nguyên
    if n == 0:
        return False
    
    # Một số là số kỳ diệu nếu tổng các chữ số cuối cùng bằng 1
    # Trong toán học, điều này tương đương với n % 9 == 1
    return n % 9 == 1

# Ví dụ kiểm tra
print(is_magic_number(10))   # True (1+0 = 1)
print(is_magic_number(1234)) # False (1+2+3+4 = 10 -> 1+0 = 1) -> True
```

Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.


```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        # Trở thành phương trình bậc nhất bx + c = 0
        if b == 0:
            return float('inf') if c == 0 else 0
        return 1
    
    delta = b**2 - 4*a*c
    
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0

# Ví dụ
print(quadratic_roots_count(1, -3, 2)) # 2 nghiệm (x=1, x=2)
```

Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    # Sử dụng công thức diện tích tam giác (Cross product)
    # Area = |x1(y2 - y3) + x2(y3 - y1) + x3(y1 - y2)| / 2
    area = x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)
    return area == 0

# Ví dụ
print(are_collinear(0, 0, 1, 1, 2, 2)) # True
    
```

Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay 
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), (0,3)], p = (2,1)



```python
def point_in_quadrilateral(quad, p):
    def sign(p1, p2, p3):
        # Kiểm tra phía của điểm p3 so với đường thẳng p1p2
        return (p3[0] - p2[0]) * (p1[1] - p2[1]) - (p1[0] - p2[0]) * (p3[1] - p2[1])

    # quad = [(x1,y1), (x2,y2), (x3,y3), (x4,y4)]
    # p = (x, y)
    
    d1 = sign(p, quad[0], quad[1])
    d2 = sign(p, quad[1], quad[2])
    d3 = sign(p, quad[2], quad[3])
    d4 = sign(p, quad[3], quad[0])

    has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0) or (d4 < 0)
    has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0) or (d4 > 0)

    # Nếu không đồng thời có cả dương và âm thì điểm nằm trong hoặc trên biên
    return not (has_neg and has_pos)

# Ví dụ từ đề bài
quad_input = [(0,0), (4,0), (4,3), (0,3)]
p_test = (2,1)
print(point_in_quadrilateral(quad_input, p_test)) # True
```

Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,
giá km, phí chờ, phụ phí ban đêm và giảm giá.


```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    """
    Tính tổng tiền taxi dựa trên:
    - car_type: '4-seats' hoặc '7-seats'
    - start_hour: Giờ bắt đầu (0-23)
    - km: Số quãng đường đi được
    - wait_minutes: Thời gian chờ (phút)
    """
    
    # 1. Cấu hình bảng giá (Có thể thay đổi tùy ý)
    config = {
        '4-seats': {
            'opening_fare': 12000,    # Giá mở cửa (cho 0.5km đầu)
            'opening_km': 0.5,
            'under_25km': 14500,     # Giá mỗi km từ sau mở cửa đến 25km
            'above_25km': 12000      # Giá mỗi km từ km thứ 26 trở đi
        },
        '7-seats': {
            'opening_fare': 15000,
            'opening_km': 0.5,
            'under_25km': 16500,
            'above_25km': 14000
        }
    }
    
    wait_fee_per_min = 1000 # Phí chờ 1000đ/phút
    night_surcharge = 0.1   # Phụ phí ban đêm 10% (từ 22h - 5h sáng)
    discount_rate = 0.05    # Giảm giá 5% nếu đi trên 50km
    
    # Lấy thông số xe
    fare_info = config.get(car_type, config['4-seats'])
    
    # 2. Tính giá theo khoảng cách (km)
    total_km_fare = 0
    if km <= fare_info['opening_km']:
        total_km_fare = fare_info['opening_fare']
    elif km <= 25:
        total_km_fare = fare_info['opening_fare'] + (km - fare_info['opening_km']) * fare_info['under_25km']
    else:
        # Tính 0.5km đầu + 24.5km tiếp theo + phần còn lại
        total_km_fare = (fare_info['opening_fare'] + 
                         (24.5 * fare_info['under_25km']) + 
                         (km - 25) * fare_info['above_25km'])
    
    # 3. Tính phí chờ
    total_wait_fare = wait_minutes * wait_fee_per_min
    
    total_fare = total_km_fare + total_wait_fare
    
    # 4. Áp dụng phụ phí ban đêm (22h đêm đến 5h sáng)
    if start_hour >= 22 or start_hour < 5:
        total_fare += total_fare * night_surcharge
        
    # 5. Áp dụng giảm giá (Ví dụ đi đường dài trên 50km)
    if km > 50:
        total_fare -= total_fare * discount_rate
        
    return round(total_fare)

# --- Chạy thử nghiệm ---
fare = smart_taxi_fare(car_type='4-seats', start_hour=23, km=30, wait_minutes=10)
print(f"Tổng tiền cước taxi của bạn là: {fare:,} VNĐ")
    
```

Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0

```python
def sum1ton(n):
    if n<=0:
        return 0
    total = 0
    for k in range(1,n+1):
        if k%11==0:
            continue
        else:
            total+=k
    return total
n = int(input())
print(sum1ton(n))
```
