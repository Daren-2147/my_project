Bài 3: Nhập vào số nguyên dương a, in ra bảng cửu chương của a

```python
# Nhập số nguyên dương a từ bàn phím
try:
    a = int(input("Nhập vào số nguyên dương a: "))

    if a > 0:
        print(f"--- Bảng cửu chương {a} ---")
        # Sử dụng vòng lặp để in từ 1 đến 10
        for i in range(1, 11):
            ket_qua = a * i
            # In theo định dạng: a x i = ket_qua
            print(f"{a} x {i} = {ket_qua}")
    else:
        print("Vui lòng nhập một số nguyên lớn hơn 0.")
        
except ValueError:
    print("Lỗi: Bạn phải nhập vào một số nguyên!")
```

Bài 4: Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím 


```python
# Nhập độ cao h từ bàn phím
h = int(input("Nhập độ cao h của tam giác cân: "))

# Duyệt qua từng hàng từ 1 đến h
for i in range(1, h + 1):
    # Số khoảng trắng ở mỗi hàng: h - i
    khoang_trang = ' ' * (h - i)
    
    # Số dấu sao ở mỗi hàng: 2*i - 1
    dau_sao = '*' * (2 * i - 1)
    
    # In hàng hiện tại
    print(khoang_trang + dau_sao)
```

Bài 10: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không.
Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 


```python
def perfect_nums(a):
    total = 0
    for k in range(1,a):
        if a%k==0:
            total += k
    if total == a:
        return True
    return False
a = int(input())
print(perfect_nums(a))
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng 

```python
# Khởi tạo các biến để lưu giá trị
so_lon_nhat = None
so_nho_nhat = None

print("Nhập vào các số nguyên (Nhập -1 để dừng):")

while True:
    try:
        n = int(input("Nhập số: "))
        
        # Nếu nhập -1 thì thoát vòng lặp ngay lập tức
        if n == -1:
            break
        
        # Nếu là số đầu tiên được nhập (các biến còn đang None)
        if so_lon_nhat is None:
            so_lon_nhat = n
            so_nho_nhat = n
        else:
            # Cập nhật số lớn nhất nếu số vừa nhập lớn hơn giá trị hiện tại
            if n > so_lon_nhat:
                so_lon_nhat = n
            
            # Cập nhật số nhỏ nhất nếu số vừa nhập nhỏ hơn giá trị hiện tại
            if n < so_nho_nhat:
                so_nho_nhat = n
                
    except ValueError:
        print("Vui lòng chỉ nhập số nguyên!")

# Sau khi thoát vòng lặp, kiểm tra xem người dùng có nhập số nào hợp lệ không
if so_lon_nhat is not None:
    print(f"Số lớn nhất là: {so_lon_nhat}")
    print(f"Số nhỏ nhất là: {so_nho_nhat}")
else:
    print("Bạn chưa nhập số nào ngoài số -1.")
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số
kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong
dãy số fibonacci lớn nhất nhưng không vượt quá A 


```python
a = int(input())
def fibonacci(n):
    if n==1 or n==2:
        return 1
    return fibonacci(n-1)+fibonacci(n-2)
k=1
while fibonacci(k)<=a:
    k+=1
print(fibonacci(k-1))
    
```

Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố

```python
def isprime(n):
    if n%2==0 and n>2 or n<2:
        return False
    for k in range(3,int(n**0.5+1),2):
        if n%k==0:
            return False
    return True
n = int(input())
count = 0
for i in range(len(str(n))):
    if isprime(int(str(n)[i])):
        count += 1
print(count)
```
