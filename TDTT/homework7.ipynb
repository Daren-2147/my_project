Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
from collections import Counter

def make_beautiful_array(arr):
    counts = Counter(arr)
    y = 0
    for x, freq in counts.items():
        if freq > x:
            y += (freq - x)  # Xóa bớt cho bằng x
        elif freq < x:
            y += freq        # Xóa hết vì không đủ x phần tử
    return y

# Test
# print(make_beautiful_array([3, 3, 3, 2, 2, 2])) -> 1 (xóa một số 2)
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì

```python
def find_majority(arr):
    candidate = None
    count = 0
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1
    
    # Kiểm tra lại xem có thực sự > n/2 không
    if arr.count(candidate) > len(arr) // 2:
        print(candidate)

# find_majority([2, 2, 1, 1, 2, 2, 2]) -> 2
```

Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; ; 

```python
def conditional_sort(arr):
    evens = sorted([x for x in arr if x % 2 == 0])
    odds = sorted([x for x in arr if x % 2 != 0], reverse=True)
    return evens + odds

# print(conditional_sort([1, 2, 3, 4, 5, 6])) -> [2, 4, 6, 5, 3, 1]
```

Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào

```python
def merge_sorted(arr1, arr2):
    # Đảm bảo cả 2 đều tăng dần để dễ xử lý
    if len(arr1) > 1 and arr1[0] > arr1[-1]: arr1.reverse()
    if len(arr2) > 1 and arr2[0] > arr2[-1]: arr2.reverse()
    
    res = []
    i = j = 0
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            res.append(arr1[i]); i += 1
        else:
            res.append(arr2[j]); j += 1
    return res + arr1[i:] + arr2[j:]
```

Cho dãy N số nguyên dương, tính tổng các số có k chữ số. Dòng đầu ghi
N số nguyên dương, dòng sau ghi số nguyên dương k

```python
def sum_k_digits(arr, k):
    return sum(x for x in arr if len(str(x)) == k)
```

Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngảy hợp lệ


```python
import datetime

def day_of_year_2020(day, month):
    try:
        date_obj = datetime.date(2020, month, day)
        return date_obj.timetuple().tm_yday
    except ValueError:
        return "Ngày không hợp lệ"
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
def common_name(names):
    # Lấy phần tên (từ cuối cùng)
    only_names = [n.split()[-1] for n in names]
    counts = Counter(only_names)
    max_freq = max(counts.values())
    # Lấy các tên có max_freq và sort để lấy tên nhỏ nhất theo từ điển
    candidates = sorted([name for name, freq in counts.items() if freq == max_freq])
    return candidates[0]
```

Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài
>=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí
tự cái.


```python
import re

def count_valid_names(s):
    # Quy tắc: Dài >=2, Hoa đầu, còn lại thường, chỉ chứa chữ cái
    pattern = r'\b[A-Z][a-z]+\b'
    return len(re.findall(pattern, s))
```

Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên

```python
def remove_adjacent_duplicates(s):
    words = s.split()
    if not words: return ""
    res = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            res.append(words[i])
    return " ".join(res)
```

Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
def max_k_digit_subsequence(s, k):
    stack = []
    to_remove = len(s) - k
    for char in s:
        while to_remove > 0 and stack and stack[-1] < char:
            stack.pop()
            to_remove -= 1
        stack.append(char)
    return "".join(stack[:k])
```

Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các
số đó. 

```python
def stats_from_text(text):
    nums = [float(x) for x in re.findall(r'\d+\.?\d*', text)]
    if not nums: return 0, 0
    return sum(nums), sum(nums)/len(nums)
```

Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
def is_valid_brackets(s):
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping:
            if not stack or stack.pop() != mapping[char]:
                return False
    return not stack
```

̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
def longest_composite_sequence(n):
    def is_prime(num):
        if num < 2: return False
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0: return False
        return True

    longest = []
    current = []
    for i in range(1, n + 1):
        if i > 1 and not is_prime(i):
            current.append(i)
        else:
            if len(current) > len(longest):
                longest = current
            current = []
    if len(current) > len(longest): longest = current
    return longest
```

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
def longest_continuous_inc(arr):
    if not arr: return []
    res = []
    temp = [arr[0]]
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            temp.append(arr[i])
        else:
            if len(temp) > len(res): res = temp
            temp = [arr[i]]
    return res if len(res) >= len(temp) else temp
```

Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím

```python
def LIS(arr):
    n = len(arr)
    if n == 0: return []
    f = [1] * n
    pre = [-1] * n
    for i in range(n):
        for j in range(i):
            if arr[i] > arr[j] and f[i] < f[j] + 1:
                f[i] = f[j] + 1
                pre[i] = j
    
    max_idx = f.index(max(f))
    res = []
    while max_idx != -1:
        res.append(arr[max_idx])
        max_idx = pre[max_idx]
    return res[::-1]
```
