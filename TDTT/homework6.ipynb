Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải


```python
n, m = map(int, input("Nhập n, m: ").split())
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

for row in matrix:
    for val in row:
        print(f"{val:>4}", end="")
    print()
```

Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input("Nhập n cho mảng vuông: "))
matrix = [list(map(int, input().split())) for _ in range(n)]

print("Đường chéo chính:", [matrix[i][i] for i in range(n)])
print("Đường chéo phụ:", [matrix[i][n-1-i] for i in range(n)])
```

Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
n, m, k = map(int, input().split())
matrix = [list(map(int, input().split())) for _ in range(n)]

tong_cot_k = sum(matrix[i][k] for i in range(n))
print(f"Tổng cột {k}: {tong_cot_k}")
```

Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def find_max_index(lst):
    if not lst: return -1
    max_val = max(lst)
    return lst.index(max_val)

# Test: print(find_max_index([1, 5, 2, 5, 3])) -> 1
```

Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def find_first_k(lst, k):
    try:
        return lst.index(k)
    except ValueError:
        return -1
```

Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
chuoi_input = input("Nhập các chuỗi: ").split()
d = {}
for idx, word in enumerate(chuoi_input):
    if word not in d:
        d[word] = []
    d[word].append(idx)

# Chuyển list vị trí thành tuple
result = {k: tuple(v) for k, v in d.items()}
print(result)
```

Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def tim_vi_tri_lon_nhat(lst):
    if not lst:  # Kiểm tra nếu danh sách rỗng
        return None
    
    # Tìm giá trị lớn nhất
    gia_tri_lon_nhat = max(lst)
    
    # Tìm vị trí (index) của giá trị đó
    vi_tri = lst.index(gia_tri_lon_nhat)
    
    return vi_tri

# Chạy thử
numbers = [10, 25, 7, 45, 30]
print(f"Vị trí phần tử lớn nhất là: {tim_vi_tri_lon_nhat(numbers)}") 
# Kết quả: 3 (vì 45 nằm ở vị trí số 3 tính từ 0)
```

VIết hàm xóa đi các số <x có trong mảng


```python
def remove_smaller_than_x(lst, x):
    # Dùng list comprehension để tạo list mới thỏa điều kiện
    return [i for i in lst if i >= x]
```

Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9


```python
def move_zeros_to_front(lst):
    zeros = [i for i in lst if i == 0]
    others = [i for i in lst if i != 0]
    return zeros + others

# Ví dụ: [0, 2, 3, 2, 4, 0, 0, 0, 9] -> [0, 0, 0, 0, 2, 3, 2, 4, 9]
```

Cho dãy chỉ gồm N giá trị từ 0 - 9 . Kiểm tra xem giá trị nào không có trong
mảng

```python
def missing_digits(lst):
    present = set(lst)
    missing = [i for i in range(10) if i not in present]
    return missing
```

Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
def sum_local_maxima(lst):
    tong = 0
    for i in range(1, len(lst) - 1):
        if lst[i] > lst[i-1] and lst[i] > lst[i+1]:
            tong += lst[i]
    return tong
```

nếu không có in ra -1

Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
def check_trend(lst):
    if len(lst) < 2: return "Vô danh"
    
    is_increasing = all(lst[i] < lst[i+1] for i in range(len(lst)-1))
    is_decreasing = all(lst[i] > lst[i+1] for i in range(len(lst)-1))
    
    if is_increasing: return "Tăng thực sự"
    if is_decreasing: return "Giảm thực sự"
    return "Vô danh"
```
