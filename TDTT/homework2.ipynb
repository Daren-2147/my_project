#### Bài 1: 
Nhập vào 2 cạnh của một hình chữ nhật. In ra màn hình diệntíchvàchu vi của nó. 

```python
# Nhập chiều dài và chiều rộng
a = float(input("Nhập chiều dài: "))
b = float(input("Nhập chiều rộng: "))

# Tính toán
dien_tich = a * b
chu_vi = (a + b) * 2

# In kết quả
print(f"Diện tích hình chữ nhật là: {dien_tich}")
print(f"Chu vi hình chữ nhật là: {chu_vi}")
```

#### Bài 2: 
Nhập vào bán kính của hình tròn. In ra màn hình diện tích và chu vi của nó. 

```python
import math

# Nhập bán kính
r = float(input("Nhập bán kính hình tròn: "))

# Tính toán
# Công thức: Chu vi = 2 * pi * r | Diện tích = pi * r^2
chu_vi = 2 * math.pi * r
dien_tich = math.pi * (r ** 2)

# In kết quả (làm tròn đến 2 chữ số thập phân)
print(f"Chu vi hình tròn là: {round(chu_vi, 2)}")
print(f"Diện tích hình tròn là: {round(dien_tich, 2)}")
```

#### Bài 3 : 
Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.


```python
import math

# Nhập 3 cạnh
a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))

# 1. Kiểm tra điều kiện là tam giác
if a + b > c and a + c > b and b + c > a:
    # 2. Phân loại tam giác
    loai_tam_giac = ""
    if a == b == c:
        loai_tam_giac = "Tam giác đều"
    elif a == b or b == c or a == c:
        if (a**2 + b**2 == round(c**2, 5)) or (a**2 + c**2 == round(b**2, 5)) or (b**2 + c**2 == round(a**2, 5)):
            loai_tam_giac = "Tam giác vuông cân"
        else:
            loai_tam_giac = "Tam giác cân"
    elif (a**2 + b**2 == round(c**2, 5)) or (a**2 + c**2 == round(b**2, 5)) or (b**2 + c**2 == round(a**2, 5)):
        loai_tam_giac = "Tam giác vuông"
    else:
        loai_tam_giac = "Tam giác thường"
    
    # 3. Tính chu vi và diện tích (Công thức Heron)
    chu_vi = a + b + c
    p = chu_vi / 2 # Nửa chu vi
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))
    
    # In kết quả
    print(f"Đây là: {loai_tam_giac}")
    print(f"Chu vi: {chu_vi}")
    print(f"Diện tích: {round(dien_tich, 2)}")
else:
    print("Khong la ba canh cua mot tam giac")
```
